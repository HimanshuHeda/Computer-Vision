# Generative Negative Transformation: Black & White Effect

This cell loads your image and applies a negative transformation to produce a black and white (inverted) effect.

```python
import cv2
import matplotlib.pyplot as plt

# Load the image (replace with your filename if needed)
img = cv2.imread('pika wallpaper.webp')
if img is None:
    raise Exception('Image not found!')

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply negative transformation (invert pixel values)
negative = 255 - gray

# Display original and negative images
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(negative, cmap='gray')
plt.title('Negative Transformation')
plt.axis('off')
plt.show()
```

# Logarithmic Transformation

This cell demonstrates how to apply a logarithmic transformation to enhance details in dark regions of an image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


# Load the image (replace with your filename if needed)

img = cv2.imread('mri.jpeg')
if img is None:
    raise Exception('Image not found!')


# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


# Apply logarithmic transformation
c = 255 / np.log(1 + np.max(gray))
log_transformed = c * np.log(1 + gray.astype(np.float32))
log_transformed = np.array(log_transformed, dtype=np.uint8)


# Display original and log transformed images
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title('Original Grayscale')
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(log_transformed, cmap='gray')
plt.title('Logarithmic Transformation')
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Step 1: Load the image
image = cv2.imread('mri.jpeg', cv2.IMREAD_GRAYSCALE)
if image is None:
    raise ValueError("❌ Image not found. Make sure 'image.jpg' is in the same folder.")

# Step 2: Convert image to float for log operation
image_float = image.astype(np.float32)

# Step 3: Logarithmic Transformation (c = 1)
c = 1
log_transformed = c * np.log(1 + image_float)

# Step 4: Normalize to 0–255
log_transformed = cv2.normalize(log_transformed, None, 0, 255, cv2.NORM_MINMAX)
log_transformed = np.uint8(log_transformed)

# Step 5: Display Original and Log Transformed Images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title('Original Grayscale Image')
plt.axis('off')
plt.imshow(image, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Logarithmic Transformed Image (c = 1)')
plt.axis('off')
plt.imshow(log_transformed, cmap='gray')

plt.tight_layout()
plt.show()

# Step 6: Save the output (optional)
cv2.imwrite('log_transformed.jpg', log_transformed)
```
