````python
import cv2

# Initialize the camera (0 for default webcam)
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not open video stream.")
else:
    ret, frame = cap.read() # Read a frame
    if ret:
        cv2.imwrite("persons.jpg", frame) # Save the captured image
        print("Image captured successfully!")
    cap.release()
    cv2.destroyAllWindows()
````

````python
import cv2

# Load the pre-trained Haar cascade for face detection
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Load an image (replace 'student_group.jpg' with your image)
img = cv2.imread('student_group.jpg')
if img is None:
    print("Error: Could not load image.")
else:
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) # Convert to grayscale for detection

    # Detect faces in the image
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5, minSize=(30, 30))

    # Draw rectangles around the detected faces
    for (x, y, w, h) in faces:
        cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2) # Blue rectangle

    # Display the result
    cv2.imshow('Detected Faces', img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
````

## Face Detection Troubleshooting & Solution

If you see "Error: Could not load image.", make sure the image file name and path are correct. The code expects an image named `student_group.jpg` in the current working directory. If your image is named differently or is in another folder, update the path accordingly.

**How to fix:**

- Place your group photo in the same folder as this notebook.
- Rename it to `student_group.jpg` or change the filename in the code to match your image.
- Example: If your image is `persons.jpg`, change this line:
```python
img = cv2.imread('student_group.jpg')
```
to
```python
img = cv2.imread('persons.jpg')
```
